# Decision Tree Classifier

```python
from sklearn.datasets import load_iris 
```

# Import iris data set using sklearn

```python
data = load_iris()
```

```python
data
```

# Importing the libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sb
```

```python
data.target
```

```python
data.feature_names
```

```python
x = data['data']
```

```python
y = data['target']
```

# Splitting the dataset into the Training set and Test set

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,random_state=42,test_size=0.3)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
model = DecisionTreeClassifier(splitter='random')
```

```python
model.fit(x_train,y_train)
```

# Display Decision Tree

# Import diabetes.csv dataset 

```python
df = pd.read_csv('diabetes.csv')
```

```python
df
```

# Check the distribution of the target

```python
df['Outcome'].value_counts()
```

```python
df.head()
```

# Replace zeros with NaN for specific columns

```python
column_to_replace = ['Glucose' , 'BloodPressure','SkinThickness','BMI','Insulin']
df[column_to_replace] = df[column_to_replace].replace(0,np.nan)
```

```python
df
```

# Check for missing values

```python
df.isnull().sum()
```

# Fill missing values with median

```python
cols = ['Glucose','BloodPressure','SkinThickness','Insulin','BMI']
```

```python
for col in cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
df.isnull().sum()
```

# Visualize Distributions

```python
sb.pairplot(df,hue='Outcome')
```

# Convert Target data into interger code

```python
df['Outcome'] = df['Outcome'].map({'Diabetic':1,'Non Diabetic' :0})
```

```python
df
```

```python
sb.heatmap(df.corr(),annot=True,cmap='coolwarm')
plt.show()
```

# Divide the data into input and output

```python
x = df.drop('Outcome',axis=1)
```

```python
x
```

```python
y = df['Outcome']
```

# Splitting the dataset into the Training set and Test set

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.3,random_state=42)
```

# Create Model

```python
model_tree = DecisionTreeClassifier()
```

# Fitting DecisionTreeClassifier   on dataset

```python
model_tree.fit(x_train,y_train)
```

# Display Decision Tree

```python
from sklearn import tree
```

```python
%matplotlib inline
```

```python
tree.plot_tree(model_tree)
```

# Predict the x_test 

```python
y_predict = model_tree.predict(x_test)
```

# Display Training Accuracy

```python
model_tree.score(x_train, y_train)
```

# Display Test Accuracy

```python
model_tree.score(x_test, y_test)
```

# Confusion Matrix

```python
from sklearn.metrics import confusion_matrix
```

```python
cm = confusion_matrix(y_test, y_predict)
```

```python
cm
```
