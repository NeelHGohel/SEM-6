# Do all pre processing on diabetes csv

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

# Import diabetes.csv data ser using sklearn

```python
df = pd.read_csv('diabetes.csv')
df
```

# Check the distribution of the target

```python
df['Outcome'].value_counts()
```

# Replace zeros with NaN for specific columns

```python
column_to_replace = ['Glucose' , 'BloodPressure','SkinThickness','BMI','Insulin']
df[column_to_replace] = df[column_to_replace].replace(0,np.nan)
```

# Check for missing values

```python
df.isnull().sum()
```

# Fill missing values with median

```python
cols = ['Glucose','BloodPressure','SkinThickness','Insulin','BMI']
```

```python
for col in cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
df.isnull().sum()
```

# Convert Targer data into interger code

```python
df['Outcome'] = df['Outcome'].map({'Non Diabetic' : 0, 'Diabetic' : 1})
df['Outcome']
```

# Visualize Distributions

```python
sns.pairplot(df,hue='Outcome')
```

```python
sns.heatmap(df.corr(), annot=True, cmap='coolwarm')
plt.show()
```

```python
df
```

# Divide the data into input and output

```python
x = df.drop('Outcome',axis=1)
```

```python
x.head()
```

```python
y = df['Outcome']
```

```python
y.head()
```

# import KNN Model

```python
from sklearn.neighbors import KNeighborsClassifier 
from sklearn.model_selection import train_test_split
```

# Convert all features into Z score

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
x_scaled = scaler.fit_transform(x)
```

```python
x_scaled
```

# Splitting the dataset into the Training set and Test set

```python
x_train, x_test,y_train,y_test = train_test_split(x_scaled,y,test_size=0.3,random_state=42)
```

# Create KNN Model

```python
model = KNeighborsClassifier(n_neighbors=100)
```

# Fitting KNN   on dataset

```python
model.fit(x_train,y_train)
```

```python
model.score(x_test, y_test)
```

# Display classification_report

```python
from sklearn.metrics import confusion_matrix,ConfusionMatrixDisplay,classification_report
```

```python
y_pred = model.predict(x_test)
```

```python
cm = confusion_matrix(y_test,y_pred)
```

```python
display = ConfusionMatrixDisplay(cm, display_labels=['Diabitic', 'Non Diabitic'])
```

```python
display.plot()
```

```python
print(classification_report(y_test,y_pred))
```

# Use Cross Validation to find best hyperparameter for KNN

```python
from sklearn.model_selection import cross_val_score
```

```python
from sklearn.model_selection import KFold
```

```python
k = KFold(n_splits=10)
```

```python
for train_index,test_index in k.split(x_scaled):
    x_train,x_test = x_scaled[train_index],x_scaled[test_index]
    y_train,y_test = y[train_index],y[test_index]
    model1 = KNeighborsClassifier()
    model1.fit(x_train,y_train)
    print(model1.score(x_test,y_test))
```

```python
cross_val_score(model,x_scaled,y,cv=k)
```
